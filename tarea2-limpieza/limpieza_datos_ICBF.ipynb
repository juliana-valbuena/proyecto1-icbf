## 1. Hallazgo: el CSV viene con un problema de formato


Al inspeccionar el archivo con `pandas.read_csv` de forma directa, columnas como
`estado_contrato` mostraban valores sin sentido (identificadores de contrato en vez de estados
como "terminado" o "Cerrado"). Se investigó a nivel de fila cruda y se encontró la causa, el
valor real de la columna **`localizaci_n`** (algo como `"Colombia, Cesar, Valledupar"`) queda
mal en la exportación y se divide en **3 tokens separados por coma en vez de 1**, por
ejemplo:

```
...,"Valledupar", <token1>Colombia , <token2> Cesar , <token3> Valledupar , "Nacional", ...
    ciudad(ok)     <---------- 3 tokens en vez de 1 (localizaci_n) ---------->  orden (desplazado)
```

Esto desplaza **+2 posiciones** todas las columnas que siguen a `localizaci_n`. El número total
de columnas por fila se mantiene en 90, por lo que el problema no es evidente contando columnas. 
Solo se detectó validando el **contenido** contra lo que cada columna debería contener.

**Validación de la corrección:** se comparó, para múltiples filas, que el contenido reubicado
tiene sentido con el nombre real de columna, fechas ISO en las columnas de fecha,
nombres de persona/organización en la columna de proveedor, códigos UNSPSC en la columna de
categoría, montos numéricos en las columnas de valor, y categorías reales y consistentes en
`estado_contrato`

**Alcance de la corrección:** se confirmó y aplicó el corrimiento de +2 para las columnas desde
`orden` (posición 9) hasta `nombre_ordenador_de_pago` (posición 83). Para las **últimas 6
columnas** (`tipo_de_documento_ordenador_de_pago` … `partition_0`) no fue posible confirmar de
forma confiable dónde exactamente se compensa el corrimiento adicional. Por lo cual, esas columnas
**se excluyen** del dataset limpio en vez de reportar valores sin verificar. Ninguna de ellas es
necesaria para las preguntas de negocio del equipo.

## 2. Carga del subconjunto de columnas confiables

Se construye el mapeo de posiciones corregido: columnas 0-7 sin cambios, `localizaci_n`
reconstruida uniendo sus 3 fragmentos, y columnas 9-83 leídas con el corrimiento +2. Las
columnas 84-89 se excluyen.

````python
# nombres reales de columna 0-83 (84 columnas), en su orden correcto en el header
NOMBRES_VALIDOS = header[0:84]

def strip_quotes(s):
    if s is None:
        return None
    s = s.strip()
    if len(s) >= 2 and s.startswith('"') and s.endswith('"'):
        s = s[1:-1]
    return s if s != "" else None

def parse_row(raw):
    # Devuelve un dict {nombre_columna_real: valor} para las 84 columnas confiables.
    if len(raw) != 90:
        return None  # fila con una corrupción distinta a la esperada -> se descarta
    rec = {}
    # columnas 0-7: sin corrimiento (renombramos las 4 columnas meta ':x' a nombres sin
    # el prefijo ':', que es solo una convención del export de Socrata/Athena)
    NOMBRES_META = {":id": "id_registro", ":version": "version_registro",
                     ":created_at": "creado_en", ":updated_at": "actualizado_en"}
    for i in range(0, 8):
        rec[NOMBRES_META.get(header[i], header[i])] = strip_quotes(raw[i])
    # localizaci_n (columna 8): reconstruida a partir de sus 3 fragmentos
    rec["localizaci_n"] = ", ".join(strip_quotes(x) or "" for x in raw[8:11]).strip(", ")
    # columnas 9-83: corrimiento +2
    for true_idx in range(9, 84):
        rec[header[true_idx]] = strip_quotes(raw[true_idx + 2])
    return rec

registros = []
descartadas = 0
with open(path, newline="", encoding="utf-8", errors="replace") as f:
    reader = csv.reader(f)
    next(reader)  # header
    for raw in reader:
        rec = parse_row(raw)
        if rec is None:
            descartadas += 1
            continue
        registros.append(rec)

df = pd.DataFrame(registros)
print(f"Filas leídas correctamente: {len(df):,}")
print(f"Filas descartadas: {descartadas}")
df.head(3)
````

## 3. Estandarización de valores faltantes

La fuente usa **varios tokens distintos** para representar "sin dato" según la columna:
`"No definido"`, `"No Definido"`, `"Sin Descripcion"`, `"No aplica"`, cadenas vacías. Se
unifican todos a `NaN`.

````python
# Diccionario de datos faltantes
FALTANTES = {"no definido", "no aplica", "sin descripcion", "sin descripción", "n/a", "na", "",}

# Columnas tipo texto, donde tiene sentido buscar estos faltantes
cols_texto = df.select_dtypes(include="object").columns

# Por cada columna de texto, se normaliza (sin espacios, minúsculas) y se reemplazan los faltantes por NaN
for c in cols_texto:
    df.loc[df[c].str.strip().str.lower().isin(FALTANTES), c] = np.nan

# % de datos faltantes por columna, ya con los placeholders convertidos
faltantes_pct = df.isna().mean().sort_values(ascending=False) * 100
print("Columnas con más datos faltantes (tras estandarizar placeholders):")
print(faltantes_pct[faltantes_pct > 0].round(1).head(20))
````

````python
# Se crea una gráfica para visualmente evidenciar que esas columnas están 100% vacías 
fig, ax = plt.subplots(figsize=(7.5, 6))
top20 = faltantes_pct[faltantes_pct > 0].head(20).iloc[::-1]
ax.barh(top20.index, top20.values, color="pink")
ax.set_xlabel("% de valores faltantes")
ax.set_title("Datos faltantes por columna (top 20, tras estandarizar placeholders)")
fig.tight_layout()
plt.show()
````

## 4. Conversión de tipos


- Columnas `fecha_*` a `datetime`.
- Columnas monetarias/numéricas (`valor_*`, `saldo_*`, `dias_adicionados`, `pilares_del_acuerdo`,
  `ultima_actualizacion`, `codigo_entidad`, `codigo_proveedor`, `espostconflicto`) a numérico.
- Se documenta un hallazgo relevante para el análisis: **`valor_del_contrato` y las demás
  columnas `valor_*` están 100% vacías** en este extracto (ya llegan así de la fuente). 
  La única columna monetaria con datos es `saldo_cdp`(Certificado de Disponibilidad Presupuestal).

````python
# Se buscan las columnas que empiezan con "fecha_" y convertirlas a tipo fecha
cols_fecha = []
for c in df.columns:
    if c.startswith("fecha_"):
        cols_fecha.append(c)

for c in cols_fecha:
    df[c] = pd.to_datetime(df[c], errors="coerce")

# Se arma la lista de columnas que deberían ser números
cols_numericas = []
for c in df.columns:
    if c.startswith("valor_") or c.startswith("saldo_"):
        cols_numericas.append(c)

otras_columnas_numericas = [
    "dias_adicionados", "pilares_del_acuerdo", "ultima_actualizacion",
    "codigo_entidad", "codigo_proveedor", "espostconflicto",
]
for c in otras_columnas_numericas:
    if c in df.columns:
        cols_numericas.append(c)

# Se convertien esas columnas a números
for c in cols_numericas:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Se revisa qué tantos datos válidos quedaron en las columnas de dinero
columnas_valor = []
for c in df.columns:
    if c.startswith("valor_"):
        columnas_valor.append(c)
columnas_valor.append("saldo_cdp")
columnas_valor.append("saldo_vigencia")

print("Cobertura de columnas monetarias (% de filas con dato):")
for c in columnas_valor:
    porcentaje_con_dato = df[c].notna().mean() * 100
    print(c, ":", round(porcentaje_con_dato, 1), "%")
````

Como se documentó antes de esta sección, las columnas de valor siguen casi todas en 0.0% de cobertura incluso después de corregir el corrimiento del CSV. Esto confirma que no era un problema de alineación, esos datos llegan vacíos desde la fuente. `saldo_cdp` es la excepción, con 81.2% de cobertura, siendo la única columna monetaria con información utilizable en este extracto.

## 5. Deduplicación

Un mismo contrato (`referencia_del_contrato`) puede tener varias filas que reflejan distintos
estados/versiones a lo largo del tiempo (columna `:version`). Nos quedamos con el registro más
reciente por contrato, usando `:actualizado_en`.

````python
antes = len(df)
n_referencias_unicas = df["referencia_del_contrato"].nunique()
duplicadas = antes - n_referencias_unicas

print("Filas totales:", antes)
print("Referencias de contrato distintas:", n_referencias_unicas)
print("Filas que son versiones repetidas de un mismo contrato:", duplicadas)

df["actualizado_en"] = pd.to_datetime(df["actualizado_en"], errors="coerce")
df = df.sort_values("actualizado_en")
df = df.drop_duplicates(subset="referencia_del_contrato", keep="last")

print("Filas después de eliminar duplicados:", len(df))
````

Se identificaron 3,199 filas correspondientes a versiones repetidas de un mismo contrato. Tras quedarnos con el registro más reciente de cada uno, el dataset pasó de 133,321 a 130,122 filas, ahora con una fila única por contrato.
