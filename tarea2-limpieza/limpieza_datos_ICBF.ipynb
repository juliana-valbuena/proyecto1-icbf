## 0. Diagnóstico inicial de datos faltantes

Primer acercamiento a los datos: carga del CSV y revisión de completitud por columna.

```python
import pandas as pd
import numpy as np
```

```python
df_ICBF = pd.read_csv("Datos ICBF.csv")
df_ICBF.head()
```

```python
df_ICBF.isna().sum()
```

```python
pct = df_ICBF.isna().mean() * 100
print(pct[pct > 10].sort_values(ascending=False).round(2))
```

**Nota sobre este diagnóstico inicial:** los porcentajes de nulos de arriba se calcularon
cargando el CSV con `pandas.read_csv` de forma directa. Al revisar de dónde salían columnas con
100% de nulos (`valor_del_contrato` y varias más) y por qué columnas como `estado_contrato` no
tenían valores reconocibles, se encontró que **el archivo fuente tiene un problema de formato**
que desalinea la mayoría de las columnas. No son datos faltantes reales, sino datos que están
en la posición equivocada.
