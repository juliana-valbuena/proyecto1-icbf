## 1. Hallazgo: el CSV viene con un problema de formato


Al inspeccionar el archivo con `pandas.read_csv` de forma directa, columnas como
`estado_contrato` mostraban valores sin sentido (identificadores de contrato en vez de estados
como "terminado" o "Cerrado"). Se investigó a nivel de fila cruda y se encontró la causa, el
valor real de la columna **`localizaci_n`** (algo como `"Colombia, Cesar, Valledupar"`) queda
mal en la exportación y se divide en **3 tokens separados por coma en vez de 1**, por
ejemplo:

```
...,"Valledupar", <token1>Colombia , <token2> Cesar , <token3> Valledupar , "Nacional", ...
    ciudad(ok)     <---------- 3 tokens en vez de 1 (localizaci_n) ---------->  orden (desplazado)
```

Esto desplaza **+2 posiciones** todas las columnas que siguen a `localizaci_n`. El número total
de columnas por fila se mantiene en 90, por lo que el problema no es evidente contando columnas. 
Solo se detectó validando el **contenido** contra lo que cada columna debería contener.

**Validación de la corrección:** se comparó, para múltiples filas, que el contenido reubicado
tiene sentido con el nombre real de columna, fechas ISO en las columnas de fecha,
nombres de persona/organización en la columna de proveedor, códigos UNSPSC en la columna de
categoría, montos numéricos en las columnas de valor, y categorías reales y consistentes en
`estado_contrato`

**Alcance de la corrección:** se confirmó y aplicó el corrimiento de +2 para las columnas desde
`orden` (posición 9) hasta `nombre_ordenador_de_pago` (posición 83). Para las **últimas 6
columnas** (`tipo_de_documento_ordenador_de_pago` … `partition_0`) no fue posible confirmar de
forma confiable dónde exactamente se compensa el corrimiento adicional. Por lo cual, esas columnas
**se excluyen** del dataset limpio en vez de reportar valores sin verificar. Ninguna de ellas es
necesaria para las preguntas de negocio del equipo.

## 2. Carga del subconjunto de columnas confiables

Se construye el mapeo de posiciones corregido: columnas 0-7 sin cambios, `localizaci_n`
reconstruida uniendo sus 3 fragmentos, y columnas 9-83 leídas con el corrimiento +2. Las
columnas 84-89 se excluyen.

````python
# nombres reales de columna 0-83 (84 columnas), en su orden correcto en el header
NOMBRES_VALIDOS = header[0:84]

def strip_quotes(s):
    if s is None:
        return None
    s = s.strip()
    if len(s) >= 2 and s.startswith('"') and s.endswith('"'):
        s = s[1:-1]
    return s if s != "" else None

def parse_row(raw):
    # Devuelve un dict {nombre_columna_real: valor} para las 84 columnas confiables.
    if len(raw) != 90:
        return None  # fila con una corrupción distinta a la esperada -> se descarta
    rec = {}
    # columnas 0-7: sin corrimiento (renombramos las 4 columnas meta ':x' a nombres sin
    # el prefijo ':', que es solo una convención del export de Socrata/Athena)
    NOMBRES_META = {":id": "id_registro", ":version": "version_registro",
                     ":created_at": "creado_en", ":updated_at": "actualizado_en"}
    for i in range(0, 8):
        rec[NOMBRES_META.get(header[i], header[i])] = strip_quotes(raw[i])
    # localizaci_n (columna 8): reconstruida a partir de sus 3 fragmentos
    rec["localizaci_n"] = ", ".join(strip_quotes(x) or "" for x in raw[8:11]).strip(", ")
    # columnas 9-83: corrimiento +2
    for true_idx in range(9, 84):
        rec[header[true_idx]] = strip_quotes(raw[true_idx + 2])
    return rec

registros = []
descartadas = 0
with open(path, newline="", encoding="utf-8", errors="replace") as f:
    reader = csv.reader(f)
    next(reader)  # header
    for raw in reader:
        rec = parse_row(raw)
        if rec is None:
            descartadas += 1
            continue
        registros.append(rec)

df = pd.DataFrame(registros)
print(f"Filas leídas correctamente: {len(df):,}")
print(f"Filas descartadas: {descartadas}")
df.head(3)
````
